```python
import numpy as np
import pandas as pd

from rdkit import Chem

from tqdm import tqdm
```

```python
def safe_mol(smiles):
    try:
        mol = Chem.MolFromSmiles(smiles, sanitize=False)
        if mol is None:
            return None

        Chem.SanitizeMol(mol)  # this is where valence/kekulization errors appear
        return mol
    except:
        return None
```

```python
df_counted = pd.read_csv("csd_counted.csv")
df_counted
```

```python
valid_rows = []
for idx, smi in df_counted["smiles"].items():
    
    if smi is np.nan:
        continue
    
    mol = safe_mol(smi)
    if mol:
        valid_rows.append(idx)

df_cleaned = df_counted.loc[valid_rows].copy()
```

```python
df_cleaned.to_csv("csd_cleaned.csv", index=False)
```
